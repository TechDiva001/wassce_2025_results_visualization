#Analyzing WASSCE 2025 Examination Results Using Data Visualization


### Project Overview:

####The West African Examinations Council (WAEC) releases annual WASSCE results that contain complex performance data across multiple subjects and years. Raw data tables make it difficult to identify trends, compare performance, and understand the scale of examination irregularities. This project transforms the official 2025 WASSCE results report into clear, actionable visual insights using data visualization techniques.

#### **Data Source**: Official WAEC press release "Release of Provisional Results for the West African Senior School Certificate Examination (WASSCE) for School Candidates, 2025" dated November 29, 2025.


####  1. IMPORT LIBRARIES

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### VISUALIZATION 1: BAR CHART - 2025 PERFORMANCE OF CORE SUBJECTS

```python
# Create DataFrame for 2025 A1-C6 percentages
performance_2025 = pd.DataFrame({
    'Subject': ['English Language', 'Mathematics', 'Integrated Science', 'Social Studies'],
    'A1-C6 Grades': [69.00, 48.73, 57.74, 55.82],
    'D7 Grade': [8.18, 11.62, 11.85, 7.38],
    'E8 Grade': [9.23, 12.15, 11.97, 9.12],
    'F9 Grade': [12.86, 26.77, 16.05, 27.5]
})
```

```python
# Create bar chart for Grades A1-C6
plt.figure(figsize=(10, 6))
sns.barplot(x=performance_2025['Subject'], y=performance_2025['A1-C6 Grades'])
plt.title("WASSCE 2025: Percentage of Candidates with Grades A1-C6")
plt.xlabel("Subject")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=45)
plt.ylim(0, 100)  # Set y-axis from 0 to 100%
plt.grid(True)
plt.show()
```

```python
# Create bar chart for Grade D7
plt.figure(figsize=(10, 6))
sns.barplot(x=performance_2025['Subject'], y=performance_2025['D7 Grade'])
plt.title("WASSCE 2025: Percentage of Candidates with Grade D7")
plt.xlabel("Subject")
plt.xticks(rotation=45)
plt.ylabel("Percentage (%)")
plt.ylim(0, 100)  # Set y-axis from 0 to 100%
plt.grid(True)
plt.show()
```

```python
# Create bar chart for Grade E8
plt.figure(figsize=(10, 6))
sns.barplot(x=performance_2025['Subject'], y=performance_2025['E8 Grade'])
plt.title("WASSCE 2025: Percentage of Candidates with Grade E8")
plt.xlabel("Subject")
plt.ylabel("Percentage (%)")
plt.ylim(0, 100)  # Set y-axis from 0 to 100%
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
# Create bar chart for Grade F9
plt.figure(figsize=(10, 6))
sns.barplot(x=performance_2025['Subject'], y=performance_2025['F9 Grade'])
plt.title("WASSCE 2025: Percentage of Candidates with Grade F9")
plt.xlabel("Subject")
plt.ylabel("Percentage (%)")
plt.ylim(0, 100)  # Set y-axis from 0 to 100%
sns.set_style("ticks")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

####  VISUALIZATION 2: LINE CHART - 4-YEAR TRENDS IN CORE SUBJECTS

```python
# Create DataFrame for trend data
trend_data = pd.DataFrame({
    'Year': [2022, 2023, 2024, 2025, 2022, 2023, 2024, 2025, 2022, 2023, 2024, 2025, 2022, 2023, 2024, 2025],
    'Subject': ['English Language', 'English Language', 'English Language', 'English Language',
                'Mathematics', 'Mathematics', 'Mathematics', 'Mathematics',
                'Integrated Science', 'Integrated Science', 'Integrated Science', 'Integrated Science',
                'Social Studies', 'Social Studies', 'Social Studies', 'Social Studies'],
    'Score': [60.39, 73.11, 69.52, 69.00,
              61.39, 62.23, 66.86, 48.73,
              62.45, 66.82, 58.77, 57.74,
              71.51, 76.76, 71.53, 55.82]
})

trend_data['Year'] = trend_data['Year'].astype(int)
```

```python

# Create line chart
plt.figure(figsize=(12, 6))
sns.lineplot(data=trend_data, x='Year', y='Score', hue='Subject')

plt.title("WASSCE Performance Trends in Core Subjects from 2022-2025 (Grades A1-C6)")
plt.xlabel("Year")
plt.ylabel("Percentage (%)")
plt.ylim(0, 100)
plt.legend(title='Subject')
plt.xticks([2022, 2023, 2024, 2025])
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

#### VISUALIZATION 3: HEATMAP - GRADE DISTRIBUTION

```python
# Set Subject as index for heatmap
grade_data_indexed = performance_2025.set_index('Subject')

# Create heatmap
plt.figure(figsize=(12, 6))
sns.heatmap(data=grade_data_indexed, annot=True, fmt='.2f')
plt.title("WASSCE 2025: Grade Distribution by Subject (%)")
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.show()
```

#### VISUALIZATION 4: BAR CHART - IRREGULARITIES

```python
# Create DataFrame for irregularities
irregularities = pd.DataFrame({
    'Category': ['Subject Results Cancelled', 'Entire Results Cancelled', 'Subject Results Withheld', 'Entire Results Withheld', 'Schools Investigated'],
    'Count': [6295, 653, 908, 158, 185]
})
```

```python
# Create bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x=irregularities['Category'], y=irregularities['Count'])
plt.title("WASSCE 2025: Examination Irregularities")
plt.xlabel("Type of Irregularity")
plt.ylabel("Number of Cases")
plt.xticks(rotation=45)  # Rotate labels for readability
plt.grid(True)
plt.show()
```

#### VISUALIZATION 5: GENDER DISTRIBUTION

```python
# Create DataFrame for gender
gender_data = pd.DataFrame({
    'Gender': ['Male', 'Female'],
    'Count': [207415, 254321]
})
```

```python
# Create bar chart (we'll simulate pie chart with bars)
plt.figure(figsize=(8, 6))
sns.barplot(x=gender_data['Gender'], y=gender_data['Count'])
plt.title("WASSCE 2025: Candidate Registration by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Candidates")
plt.grid(True)
plt.show()
```

#### VISUALIZATION 6: SCATTER PLOT - SUBJECT RELATIONSHIP

```python
# Let's compare Math vs Science performance
# We'll use the 4-year data to see if they correlate

# Filter for Math and Science
math_science_data = trend_data[trend_data['Subject'].isin(['Mathematics', 'Integrated Science'])]

# Filter for English and Social Studies
english_social_studies_data = trend_data[trend_data['Subject'].isin(['English Language', 'Social Studies'])]

# Pivot to get Math and Science as separate columns
pivot_data1 = math_science_data.pivot(index='Year', columns='Subject', values='Score').reset_index()

# Pivot to get English and Social Studies as separate columns
pivot_data2 = english_social_studies_data.pivot(index='Year', columns='Subject', values='Score').reset_index()
```

```python
# Create scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x=pivot_data1['Mathematics'], y=pivot_data1['Integrated Science'])
plt.title("Relationship: Math vs Science Performance (2022-2025)")
plt.xlabel("Mathematics Score (%)")
plt.ylabel("Science Score (%)")
plt.xlim(40, 70)
plt.ylim(50, 70)
plt.grid(True)

# Add regression line to see trend
sns.regplot(x=pivot_data1['Mathematics'], y=pivot_data1['Integrated Science'], scatter=False, color='red')
plt.show()
```

**Conclusion from Scatter Plot (Maths vs Science)**

The scatter plot and regression line show a moderate positive correlation between Mathematics and Science performance from 2022–2025. As Mathematics scores increase, Science scores also tend to rise. Science consistently performs slightly better than Mathematics across all years. This trend suggests that improvements in Mathematics instruction or student ability may also influence Science performance, since the two subjects show similar learning patterns.

```python
# Create scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x=pivot_data2['English Language'], y=pivot_data2['Social Studies'])
plt.title("Relationship: English vs Social Studies Performance (2022-2025)")
plt.xlabel("English Score (%)")
plt.ylabel("Social Studies Score (%)")
plt.xlim(40, 70)
plt.ylim(50, 70)
plt.grid(True)

# Add regression line to see trend
sns.regplot(x=pivot_data2['English Language'], y=pivot_data2['Social Studies'], scatter=False, color='red')
plt.show()
```

**Conclusion from Scatter Plot (English vs Social Studies)**

The scatter plot and regression line show a moderate positive correlation between English and Social Studies performance from 2022–2025. As English scores increase, Social Studies scores also tend to rise. Social Studies consistently performs slightly better than English across the observed score ranges. This trend suggests that strong language comprehension and analytical skills developed in English positively influence performance in Social Studies, indicating shared cognitive and literacy-based foundations between the two subjects.
